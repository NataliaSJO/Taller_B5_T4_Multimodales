# Cargar el modelo

Utilizaremos la API `pipeline` de 🤗 Transformers para simplificar la interacción con el modelo: **texto → modelo generativo → señal de audio**

La pipeline se encarga de buena parte del trabajo necesario para transformar nuestro texto en una representación de audio.

En este caso utilizamos:

- `text-to-speech`: indica la tarea.
- `suno/bark`: modelo que realizará la generación.

```python
from transformers import pipeline

synthesizer = pipeline("text-to-speech", "suno/bark")
```

¿Qué ocurre internamente? Conceptualmente, podemos pensar en el proceso como:

**Texto → representación interna → generación de audio → waveform**

La salida no es simplemente un archivo de sonido. La pipeline devuelve información que podemos utilizar para reproducir o guardar el audio.

Esto nos permite separar dos conceptos:

1. **Generación:** el modelo crea la señal de audio.
2. **Reproducción:** Python/Jupyter se encarga de presentarla para que podamos escucharla.

Vamos a empezar con un ejemplo muy sencillo. La variable `output` contendrá el resultado generado por el modelo.

```python
output = synthesizer("He generado un audio de forma muy sencilla para la clase de Pablo")
```

El resultado contiene, entre otros elementos, el **audio** y su **frecuencia de muestreo** (`sampling_rate`). La frecuencia de muestreo indica cuántas muestras de la señal se utilizan por segundo para representar el sonido digital.

```python
from IPython.display import Audio
Audio(output["audio"], rate=output["sampling_rate"])
```

## Del texto hablado a una interpretación más expresiva

Los modelos generativos de audio pueden responder a determinadas convenciones incluidas en el texto. Por ejemplo, podemos utilizar símbolos musicales (`♪`) para intentar inducir una interpretación cantada.

Esto es interesante porque nos muestra una característica importante de los modelos generativos: **El prompt no siempre describe únicamente el contenido; también puede proporcionar pistas sobre cómo queremos que se genere.**


```python
output = synthesizer("♪ He generado un audio de forma muy sencilla para la clase de Pablo ♪")
Audio(output["audio"], rate=output["sampling_rate"])
```

El siguiente ejemplo introduce texto en inglés y una indicación entre corchetes. Vamos a ver cómo cambia la interpretación del modelo.

```python
output = synthesizer("Hello, my name is Suno. And, uh — and I like pizza. [laughs] But I also have other interests such as playing tic tac toe.")
Audio(output["audio"], rate=output["sampling_rate"])
```
